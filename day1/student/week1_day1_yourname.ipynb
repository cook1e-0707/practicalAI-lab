# Mean Absolute Error (MAE)

A model makes predictions, but predictions are rarely perfect. We need
one number that tells us how far the predictions are from the real
answers.

**MAE** means **Mean Absolute Error**.

In this project, MAE answers:

> On average, about how many cups away are the model's predictions?

Consider three days:

| Day | Real sales | Predicted sales | Difference |
|---:|---:|---:|---:|
| 1 | 80 cups | 75 cups | 5 cups |
| 2 | 60 cups | 68 cups | 8 cups |
| 3 | 100 cups | 94 cups | 6 cups |

The average difference is:

```text
(5 + 8 + 6) / 3 = 6.3 cups
```

We ignore whether a prediction is too high or too low. We only measure
how far away it is.


## MAE Example

The function `mean_absolute_error(...)` compares the real values with
the predicted values. It also comes from scikit-learn, so we import it
immediately before using it.

Read the code, predict the output, and then run it.


````python
from sklearn.metrics import mean_absolute_error

actual_sales = [80, 60, 100]
predicted_sales = [75, 68, 94]

example_mae = mean_absolute_error(
    actual_sales,
    predicted_sales,
)

print("Example MAE:", round(example_mae, 1), "cups")
````

## What MAE Means

If the output is `6.3 cups`, say:

> The predictions are typically about 6.3 cups away from the real
> sales.

Remember:

- a **smaller** MAE is better;
- MAE equal to `0` means every prediction is correct;
- MAE uses the same unit as the answer—in this project, **cups**;
- MAE is not a percentage or an accuracy score.

# Model A: Learned Line

`.fit()` does not memorize every blue point. It learns two numbers that
describe one straight line:

```text
predicted cups = intercept + slope × temperature
```

The next cell reads those two learned numbers from Model A. This code is
provided; you do not need to memorize `coef_` or `intercept_`.


````python
slope = model_A.coef_[0]
intercept = model_A.intercept_

print("Learned slope:", round(slope, 2))
print("Learned intercept:", round(intercept, 2))
print(
    "Prediction rule:",
    round(intercept, 2),
    "+",
    round(slope, 2),
    "× temperature",
)
````

## Slope

If the learned slope is about `1.3`, the model is saying:

> For each 1°F increase in temperature, predicted sales increase by
> about 1.3 cups.

The intercept helps position the line. It describes the model's
prediction at 0°F, but 0°F is outside this dataset, so we should not give
that number a strong real-world interpretation.


## Model A on Training Data

- Blue circles are the real training values.
- Red crosses are the model's predictions.
- The red crosses form a straight line because this is linear
  regression.


````python
training_predictions = model_A.predict(
    train_data[features_A]
)

plt.scatter(
    train_data["temperature_f"],
    train_data["cups_sold"],
    color="#1976D2",
    label="Real training sales",
)
plt.scatter(
    train_data["temperature_f"],
    training_predictions,
    color="#E53935",
    marker="x",
    s=70,
    label="Model predictions",
)
plt.xlabel("Temperature (°F)")
plt.ylabel("Cups sold")
plt.title("The line learned from training data")
plt.legend()
plt.show()
````
