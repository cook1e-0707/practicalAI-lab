## String and `str` Are Not the Same Spelling

**String** is the full English name of the idea. Python's built-in type
name is lowercase **`str`**.

A string is commonly created in two ways:

```python
color = "Blue"       # Create a string by writing text in quotes.
number_text = str(12) # Convert the number 12 to the string "12".
```

Python is **case-sensitive**, which means uppercase and lowercase
letters are treated as different names:

```text
str       correct built-in Python type name
String    not the built-in string type
STR       not the built-in string type

True      correct Boolean value
true      a different name, not Python's Boolean value
```

Therefore, Python normally uses `"Blue"` or `str(...)`, not
`String(...)`. Other programming languages may use `String`, which is
why you may have seen that spelling elsewhere.
